driver-assistance applications, and educational demonstrations of
computer vision.

## 5. Introduction to EfficientNet-B0

EfficientNet-B0 is a convolutional neural network (CNN) architecture
designed for image classification. It is the smallest model in the
EfficientNet family and provides a good balance between accuracy and
computational efficiency.

In this project, a pre-trained EfficientNet-B0 model is used to
classify road-sign images. Instead of training a deep learning model
from scratch, the already trained model is used to perform inference
on input images.

The model accepts an image as input after suitable preprocessing and
produces prediction scores for different image classes. The class
with the highest prediction score is selected as the predicted output.

## 6. Why EfficientNet-B0 Was Selected

EfficientNet-B0 was selected because it is a lightweight and efficient
image classification model. It provides a good balance between model
size, computational requirements, and classification performance.

Using a pre-trained model also reduces the need to train a deep
learning model from scratch. This makes it suitable for demonstrating
the practical use of pre-trained models in image classification.

```python
from torchvision.datasets import GTSRB

data_path = "./database"

train_dataset = GTSRB(
    root=data_path,
    split="train",
    download=False
)

print("Dataset loaded successfully!")
print("Number of training images:", len(train_dataset))
```

```python
# Find the number of road-sign classes

labels = [sample[1] for sample in train_dataset._samples]
num_classes = len(set(labels))

print("Number of road-sign classes:", num_classes)
```

```python
from torchvision.models import efficientnet_b0, EfficientNet_B0_Weights

# Load the pretrained EfficientNet-B0
weights = EfficientNet_B0_Weights.DEFAULT
model = efficientnet_b0(weights=weights)

# Check the current classifier
print(model.classifier)
```

```python
import torch.nn as nn

# Replace the final classifier for 43 road-sign classes
model.classifier[1] = nn.Linear(
    in_features=1280,
    out_features=43
)

print("EfficientNet-B0 classifier updated successfully!")
print(model.classifier)
```

```python
# Freeze all pretrained EfficientNet-B0 layers
for param in model.features.parameters():
    param.requires_grad = False

# Keep the new classifier trainable
for param in model.classifier.parameters():
    param.requires_grad = True

print("Pretrained layers frozen.")
print("New classifier is trainable.")
```

```python
# Get the recommended preprocessing for pretrained EfficientNet-B0
weights = EfficientNet_B0_Weights.DEFAULT
preprocess = weights.transforms()

print("Preprocessing pipeline created successfully!")
print(preprocess)
```

```python
# Create a transformed version of the training dataset
train_dataset_processed = GTSRB(
    root="./database",
    split="train",
    transform=preprocess,
    download=False
)

print("Preprocessed dataset created successfully!")
print("Number of training images:", len(train_dataset_processed))
```

```python
from torch.utils.data import DataLoader

# Create a DataLoader for the training dataset
train_loader = DataLoader(
    train_dataset_processed,
    batch_size=32,
    shuffle=True,
    num_workers=0
)

print("Training DataLoader created successfully!")
print("Number of batches:", len(train_loader))
```

```python
# Get one batch of images and labels
images, labels = next(iter(train_loader))

print("Batch loaded successfully!")
print("Image batch shape:", images.shape)
print("Label batch shape:", labels.shape)
print("First 10 labels:", labels[:10])
```

```python
import torch

# Check whether a GPU is available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using device:", device)
```

```python
# Move the model to the selected device
model = model.to(device)

print("Model moved to:", device)
```

```python
import torch.nn as nn
import torch.optim as optim

# Loss function for multi-class classification
criterion = nn.CrossEntropyLoss()

# Optimizer for the trainable classifier
optimizer = optim.Adam(
    model.classifier.parameters(),
    lr=0.001
)

print("Loss function and optimizer created successfully!")
```

```python
# Train the new classifier for 1 epoch

model.train()

running_loss = 0.0

for batch_idx, (images, labels) in enumerate(train_loader):

    # Move data to CPU
    images = images.to(device)
    labels = labels.to(device)

    # Clear previous gradients
    optimizer.zero_grad()

    # Forward pass
    outputs = model(images)

    # Calculate loss
    loss = criterion(outputs, labels)

    # Backward pass
    loss.backward()

    # Update classifier weights
    optimizer.step()

    running_loss += loss.item()

    # Show progress every 100 batches
    if (batch_idx + 1) % 100 == 0:
        print(
            f"Batch [{batch_idx + 1}/{len(train_loader)}], "
            f"Loss: {running_loss / (batch_idx + 1):.4f}"
        )

print("Training completed successfully!")
```

```python
# Load the GTSRB test dataset with the same preprocessing

test_dataset = GTSRB(
    root="./database",
    split="test",
    transform=preprocess,
    download=True
)

print("Test dataset loaded successfully!")
print("Number of test images:", len(test_dataset))
```

```python
from torch.utils.data import DataLoader

# Create DataLoader for the test dataset
test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=0
)

print("Test DataLoader created successfully!")
print("Number of test batches:", len(test_loader))
```

```python
# Evaluate the model on a smaller test set
# This is faster on CPU

model.eval()

correct = 0
total = 0

max_batches = 50

with torch.no_grad():
    for batch_idx, (images, labels) in enumerate(test_loader):

        if batch_idx >= max_batches:
            break

        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total

print("Test evaluation completed!")
print("Test Accuracy on sampled data: {:.2f}%".format(accuracy))
print("Correct predictions:", correct)
print("Total images evaluated:", total)
```

```python
# Save the trained EfficientNet-B0 model
model_path = "./database/efficientnet_b0_road_signs.pth"

torch.save(model.state_dict(), model_path)

print("Trained model saved successfully!")
print("Saved at:", model_path)
```

```python
# Select 3 different test images for prediction

test_indices = [0, 100, 500]

print("Selected test cases:")
for i, idx in enumerate(test_indices, start=1):
    _, actual_label = test_dataset[idx]
    print(f"Test Case {i}: Image index = {idx}, Actual class = {actual_label}")
```

```python
# Run inference on the 3 selected test cases

model.eval()

print("Prediction Results")
print("-" * 40)

with torch.no_grad():
    for i, idx in enumerate(test_indices, start=1):
        image, actual_label = test_dataset[idx]

        # Add batch dimension and move image to device
        image = image.unsqueeze(0).to(device)

        # Model inference
        output = model(image)

        # Get predicted class
        predicted_label = torch.argmax(output, dim=1).item()

        print(f"Test Case {i}")
        print(f"Image Index   : {idx}")
        print(f"Actual Class  : {actual_label}")
        print(f"Predicted Class: {predicted_label}")
        print("-" * 40)
```

```python
# Visualize predictions for the 3 test cases

import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

model.eval()

with torch.no_grad():
    for i, idx in enumerate(test_indices):
        image, actual_label = test_dataset[idx]

        # Model prediction
        input_image = image.unsqueeze(0).to(device)
        output = model(input_image)
        predicted_label = torch.argmax(output, dim=1).item()

        # Convert image for display
        display_image = image.permute(1, 2, 0).numpy()

        # Undo normalization
        mean = np.array([0.485, 0.456, 0.406])
        std = np.array([0.229, 0.224, 0.225])

        display_image = display_image * std + mean
        display_image = np.clip(display_image, 0, 1)

        axes[i].imshow(display_image)
        axes[i].set_title(
            f"Actual: {actual_label}\nPredicted: {predicted_label}"
        )
        axes[i].axis("off")

plt.tight_layout()
plt.show()
plt.close()
```

```python
import torch
import torch.nn as nn
from torchvision.models import efficientnet_b0, EfficientNet_B0_Weights

# Load EfficientNet-B0 architecture
weights = EfficientNet_B0_Weights.DEFAULT
model = efficientnet_b0(weights=weights)

# Change classifier to 43 road-sign classes
model.classifier[1] = nn.Linear(
    in_features=1280,
    out_features=43
)

# Load our trained classifier weights
model.load_state_dict(
    torch.load(
        "./database/efficientnet_b0_road_signs.pth",
        map_location="cpu"
    )
)

# Set model to evaluation mode
model.eval()

# Use CPU
device = torch.device("cpu")
model = model.to(device)

print("Trained road-sign model loaded successfully!")
```

```python
from torchvision.datasets import GTSRB

# Preprocessing for EfficientNet-B0
weights = EfficientNet_B0_Weights.DEFAULT
preprocess = weights.transforms()

# Load the GTSRB test dataset
test_dataset = GTSRB(
    root="./database",
    split="test",
    transform=preprocess,
    download=False
)

# Our 3 test cases
test_indices = [0, 100, 500]

print("Test dataset loaded successfully!")
print("Number of test images:", len(test_dataset))
print("Test indices:", test_indices)
```

```python
import numpy as np
import matplotlib.pyplot as plt

model.eval()

fig, axes = plt.subplots(1, 3, figsize=(9, 3))

with torch.no_grad():
    for i, idx in enumerate(test_indices):

        image, actual_label = test_dataset[idx]

        # Prediction
        input_image = image.unsqueeze(0)
        output = model(input_image)
        predicted_label = torch.argmax(output, dim=1).item()

        # Convert normalized tensor back to displayable image
        display_image = image.permute(1, 2, 0).numpy()

        mean = np.array([0.485, 0.456, 0.406])
        std = np.array([0.229, 0.224, 0.225])

        display_image = display_image * std + mean
        display_image = np.clip(display_image, 0, 1)

        # Display
        axes[i].imshow(display_image)
        axes[i].set_title(
            f"Actual: {actual_label}\nPredicted: {predicted_label}"
        )
        axes[i].axis("off")

plt.tight_layout()
plt.show()
plt.close()
```

- Only the classifier layer was trained, while the pretrained feature extraction layers were kept frozen.
